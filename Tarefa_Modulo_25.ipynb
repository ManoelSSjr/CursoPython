```python
import sqlite3 #Importando a biblioteca necessária
import pandas as pd
```

```python
df_vendas = pd.read_csv("C:/Users/user/Documents/EBAC/dados/TB_VENDAS_TAREFA.csv", delimiter=';')
```

```python
# Criar uma conexao com um banco de dados SQLite em memoria
conn = sqlite3.connect(':memory:')

# Carregar o dataframe no banco de dados do SQLite
df_vendas.to_sql('tb_vendas', conn, index=False, if_exists='replace')
```

```python
# funcao para executar consultas SQL e retornar o resultado como um dataframe
def run_query(query):
    return pd.read_sql_query(query, conn)
```

```python
# 1) Faça uma consulta que retorne os nomes dos produtos distintos que temos na base de venda.
```

```python
query = """
SELECT
DISTINCT(PRODUTO)
from tb_vendas
"""
result_df = run_query(query)
print(result_df)
```

```python
# 2) Faça uma consulta que retorne a contagem dos clientes distintos que temos na nossa base, não esqueça de renomear sua coluna.

# Dica: Você pode usar o distinct dentro do count()
```

```python
query = """
SELECT
COUNT(DISTINCT ID_CLIENTE) AS total_clientes_distintos
FROM tb_vendas;
"""
result_df = run_query(query)
print(result_df)
```

```python
# 3) Faça uma consulta que retorne uma coluna com os produtos distintos e o valor_unid de cada produto. Porém apenas para produtos onde o valor_unid é 
# maior ou igual a 50 reais.
```

```python
query = """
SELECT 
DISTINCT PRODUTO, VALOR_UNID
FROM tb_vendas
WHERE VALOR_UNID >= 50;
"""
result_df = run_query(query)
print(result_df)
```

```python
# [DESAFIO] 4) Faça uma consulta que retorne o ID das compras e o valor total gasto nessa compra (Valor_unidade X Unidades) das 5 compras com mais 
# valor total gasto.
```

```python
query = """
SELECT 
ID_COMPRA, (VALOR_UNID * UNIDADES) AS VALOR_TOTAL_GASTO
FROM tb_vendas
ORDER BY VALOR_TOTAL_GASTO DESC
LIMIT 5;
"""
result_df = run_query(query)
print(result_df)
```

```python
# 5) Faça uma consulta que retorne os produtos e a média do preço da unidade dos produtos, ordenando do maior para o menor.
```

```python
query = """
SELECT 
PRODUTO, AVG(VALOR_UNID) AS media_preco_unitario
FROM tb_vendas
GROUP BY PRODUTO
ORDER BY media_preco_unitario DESC;
"""
result_df = run_query(query)
print(result_df)
```

```python
# 6) Faça uma consulta que retorne o id dos 3 clientes da base de vendas que mais relalizaram compras e a quantidade de compras realizadas.
```

```python
query = """
SELECT 
ID_CLIENTE, COUNT(DISTINCT ID_COMPRA) AS quantidade_compras
FROM tb_vendas
GROUP BY ID_CLIENTE
ORDER BY quantidade_compras DESC
LIMIT 3;
"""
result_df = run_query(query)
print(result_df)
```
