```python
import sqlite3 #Importando a biblioteca necessária
import pandas as pd
```

```python
df_vendas = pd.read_csv("C:/Users/user/Documents/EBAC/dados/TB_VENDAS_TAREFA.csv", delimiter=';')
```

```python
# Criar uma conexao com um banco de dados SQLite em memoria
conn = sqlite3.connect(':memory:')

# Carregar o dataframe no banco de dados do SQLite
df_vendas.to_sql('tb_vendas', conn, index=False, if_exists='replace')
```

```python
# funcao para executar consultas SQL e retornar o resultado como um dataframe
def run_query(query):
    return pd.read_sql_query(query, conn)
```

```python
# 1) Faça uma consulta que retorne todos os dados da nossa tabela de vendas, com todas linhas e todas colunas.
```

```python
query = "SELECT * from tb_vendas"
result_df = run_query(query)
print(result_df)
```

```python
# 2) Faça uma consulta que retorne apenas 10 linhas da coluna produto.
```

```python
query = "SELECT * from tb_vendas limit 10"
result_df = run_query(query)
print(result_df)
```

```python
# 3) Faça uma consulta que retorne a média do valor por unidade e também das unidades vendidas. Não se esqueça de renomear o nome das colunas ao cria-las.
```

```python
query = "SELECT AVG(VALOR_UNID) AS MEDIA_VALOR_UNID, AVG(UNIDADES) AS MEDIA_UNIDADES from tb_vendas"
result_df = run_query(query)
print(result_df)
```

```python
# 4) Note que na nossa tabela não temos o total de valor gasto por compra, temos o valor do produto e a quantidade vendida, para termos o valor total 
# precisamos multiplicar essas duas colunas e para multiplicar colunas usamos o *.
# Faça uma consulta que retorne a coluna ID_COMPRA, ID_CLIENTE e o valor total gasto.
# Ex: SELECT TICKET_MEDIO*UNIDADE FROM TB_EXEMPLO
```

```python
query = "SELECT (VALOR_UNID*UNIDADES) AS VALOR_GASTO from tb_vendas"
result_df = run_query(query)
print(result_df)
```

```python
# 5) [DESAFIO] Faça uma consulta que traz a média do valor total gasto que foi calculado no exercício anterior. 
# Note que não queremos a média por pedido e sim a média total, apenas essa informação deve vir na consulta.
```

```python
query = "SELECT AVG(VALOR_UNID * UNIDADES) AS MEDIA_VALOR_GASTO FROM tb_vendas"
result_df = run_query(query)
# Exibir o resultado
print(result_df)
```
