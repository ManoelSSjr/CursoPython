```python
# 2 - Para essa base, onde você realizará as previsões de fire alarm, qual modelo de machine learning você aplicará? Justifique.
```

```python
# Utilizaremos regressão logística, para conseguir fazer a classificao binaria. Para uma variável binária, o objetivo é estimar a probabilidade de 
# ocorrência de um dos dois resultados (geralmente codificado como 1).
```

```python
# 3 - Separe a base em Y e X e já rode a instância do modelo que você utilizará.
```

```python
X = df.drop('Fire_Alarm', axis=1)
y = df['Fire_Alarm']
```

```python
# Instanciando o modelo de regressão logística
modelo = LogisticRegression(max_iter=1000)
```

```python
# 4 - Defina o número de Folds e rode o modelo com a validação cruzada.
```

```python
folds = 5
```

```python
crossvalidation = KFold(n_splits=folds, shuffle=True, random_state=5)
```

```python
modelo_final = cross_val_score(modelo, X, y, cv = folds)
```

```python
# 5 - Avalie a pontuação de cada modelo e ao final a validação final da média.
```

```python
pontuacoes = cross_val_score(modelo, X, y, cv=crossvalidation)

print(f"Pontuações por fold: {pontuacoes}")
```

```python
#  Esta linha imprime as pontuações obtidas em cada uma das 5 divisões (folds). Cada valor representa a performance do modelo em um fold específico, 
# onde o modelo foi treinado com 4 dos 5 folds e testado no fold restante. Essas pontuações são obtidas através da métrica padrão usada pelo 
# cross_val_score, que geralmente é a acurácia para problemas de classificação.
```

```python
# Os valores de acurácia obtidos sao de aproximadamente 97,6% 97,5% 98,8% 98,62% e 97,6%
```
