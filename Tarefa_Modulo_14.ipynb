```python
# boxplor da coluna Pagamento_Mensal

plt.figure(figsize=(8, 6))
sns.boxplot(data=df, y='Pagamento_Mensal', color='skyblue')
plt.title('Boxplot da coluna do pagamento mensal')
plt.ylabel('Pagamento mensal')
plt.show()
```
