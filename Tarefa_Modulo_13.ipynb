```python
import pandas as pd
import matplotlib.pyplot as plt
import plotly.express as px
import seaborn as sns
import plotly.express as px
```

```python
df = pd.read_csv('C:/Users/user/Documents/EBAC/dados/MODULO7_PROJETOFINAL_BASE_SUPERMERCADO_-_MODULO7_PROJETOFINAL_BASE_SUPERMERCADO_(1).csv.csv')

df.head(10)
```

```python
# 1 - Traga a média e a mediana dos preços - coluna Preco_Normal - por categoria de produto.
# Identifique as categorias que parecem ter um valor de média abaixo ou acima da mediana.
```

```python
# Media

df['Preco_Normal'].mean()
```

```python
# Media por categoria 

df.groupby('Categoria')['Preco_Normal'].mean().reset_index().sort_values(by="Categoria", ascending=False)
```

```python
#Mediana 

df['Preco_Normal'].median()
```

```python
# Mediana por categoria

df.groupby('Categoria')['Preco_Normal'].median().reset_index().sort_values(by="Categoria", ascending=False)
```

```python
# A categoria comidas-preparadas, é a única que possui valor da mediana maior doque sua própria média.
```

```python
# 2 - Traga o desvio padrão por categoria de produto.
# Qual o comportamento da média e mediana nas categorias com maior desvio?
```

```python
# Desvio padrão por categoria 

desvio_padrao_por_categoria = df.groupby('Categoria')['Preco_Normal'].std().reset_index()
desvio_padrao_por_categoria
```

```python
# As categorias belleza-y-cuidado-personal, instantaneos-y-sopas, lacteos estão com o desvio padrao acima dos valores de media e mediana, demonstrando que alguns produtos estão muito acima da média e mediana, enquanto que outros produtos estão bem abaixo.

# A categoria comidas-preparadas apresenta desvio padrao muito abaixo da media e da mediana, assim é possivel afirmar que existe pouca variacao de preco dos produtos desta categoria

# A categoria frutas apresenta desvio padrao proximo a media e bastante acima da mediana, demonstrando que alguns produtos distoam com preços mais altos

# A categoria congelados apresenta um desvio padrão acima da mediana, porem quase que empatado com a média, demonstrando equilibrio entre o preco nos produtos

# A categoria frutas apresenta desvio padrao abaixo da mediana e mais abaixo ainda da média, demonstrando pouca movimentação de preco entre os produtos
```

```python
# 3 - Plot um boxplot da distribuição do Preco_Normal para a categoria que você identificou que tem o maior desvio padrão. Como é a distribuição desses 
# dados segundo o boxplot? Você identifica muitos outliers?
```

```python
dados_belleza = df.loc[df['Categoria'] == 'belleza-y-cuidado-personal']
dados_belleza
```

```python
#Boxplot

plt.figure(figsize=(10, 6))  # Define o tamanho da figura
sns.boxplot(x=dados_belleza['Preco_Normal'])
plt.title(f'Boxplot de Preco_Normal para a categoria {dados_belleza}')
plt.xlabel('Preco_Normal')
plt.show()
```

```python
# É possivel identificar outliers em valores elevados da variavel preco_normal
```

```python
# 4 - Plote um gráfico de barras onde temos a média de descontos por categoria.
```

```python
# Gráfico de barras

# Criando a figura do gráfico
plt.figure(figsize=(10, 6))

# Agrupando os dados por profissão e calculando a média dos salários
salario_por_profissao = df.groupby("Categoria")["Desconto"].mean()

# Criando o gráfico de barras
plt.bar(salario_por_profissao.index, salario_por_profissao, color="skyblue")

# Corrigindo o rótulo do eixo x
plt.xlabel("Categoria", fontsize=12)
plt.ylabel("Desconto", fontsize=12)
plt.title("Média de descontos por categoria", fontsize=16)

# Melhorando a legibilidade dos rótulos do eixo x
plt.xticks(rotation=45)

# Adicionando grades para facilitar a leitura dos valores
plt.grid(True, linestyle="--", alpha=0.7)

# Exibindo o gráfico
plt.show()
```

```python
# 5 - Plote um gráfico de mapa interativo agrupando os dados por categoria, marca e trazendo a média de desconto.
```

```python
# media de desconto por categoria:

# Calcula a média de desconto por categoria, marca
df_grouped = df.groupby(['Categoria', 'Marca'])['Desconto'].mean().reset_index()

# Cria o gráfico de mapa interativo
fig = px.scatter_mapbox(df_grouped,
                        color='Categoria',
                        size='Desconto',
                        hover_name='Marca',
                        hover_data=['Categoria', 'Desconto'],
                        zoom=5,
                        mapbox_style='carto-positron')

# Exibe o gráfico
fig.show()
```
